```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

df_x, y_series = load_breast_cancer(as_frame=True, return_X_y=True)

y = pd.DataFrame(y_series)

print("First 10 Records:")
display(df_x.head(10))

print("Dataset Shape:")
print(df_x.shape)

print("\nFeature Names:")
print(df_x.columns.tolist())

print("\nTarget Classes:")
print(y_series.unique())
```

### Q4. Data Preprocessing

```python
# Separating features (X) and target (y)

X = df_x
y = y_series

print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)
```

```python
# Splitting the data into 80% training and 20% testing sets

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
print("Training Class Distribution:")
print(y_train.value_counts())

print("\nTesting Class Distribution:")
print(y_test.value_counts())
```

### Q5. Model Training

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=10000)
model.fit(X_train, y_train)
```

```python
print("Model Coefficients:")
print(model.coef_)
```

```python
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_[0]
})

display(coefficients)
```

```python
print("Model Intercept:")
print(model.intercept_)
```

### Q6. Prediction

```python
y_pred = model.predict(X_test)

print("Predicted Values:")
print(y_pred)
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

display(comparison)
```

```python
correct = (y_test.values == y_pred).sum()
incorrect = (y_test.values != y_pred).sum()

print("Correct Predictions:", correct)
print("Incorrect Predictions:", incorrect)
```

### Q7. Model Evaluation

```python
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,classification_report)
```

```python
# Accuracy Score
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
# Precision Score
precision = precision_score(y_test, y_pred)

print("Precision:", precision)
```

```python
# Recall Score
recall = recall_score(y_test, y_pred)

print("Recall:", recall)
```

```python
# F1-Score
f1 = f1_score(y_test, y_pred)

print("F1-Score:", f1)
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
# Classification Report
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

### Q8. Model Visualization

```python
# Confusion Matrix Heatmap

plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Malignant", "Benign"],
    yticklabels=["Malignant", "Benign"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.show()
```

### Q9. Feature Analysis

```python
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_[0]
})

feature_importance["Absolute_Coefficient"] = (
    feature_importance["Coefficient"].abs()
)

display(feature_importance)
```

```python
top_5_features = feature_importance.sort_values(
    by="Absolute_Coefficient",
    ascending=False
).head(5)

display(top_5_features)
```

```python
top_5_features = top_5_features.copy()

top_5_features["Influence"] = top_5_features["Coefficient"].apply(
    lambda x: "Towards Benign (1)" if x > 0 else "Towards Malignant (0)"
)

display(top_5_features)
```
