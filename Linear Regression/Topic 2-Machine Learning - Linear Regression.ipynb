```python
# Find the most correlated feature
correlation_with_target = df.corr()['target'].drop('target').abs().sort_values(ascending=False)

print("Feature ranked by absolute correlation with target :")
print(correlation_with_target)
```

```python
# Get most corelated Features 
most_corrlated_feature = correlation_with_target.index[0]

print("Most correlated features :", most_corrlated_feature)

#scatter plot
plt.figure(figsize=(6,4))

sns.scatterplot(
    x=df[most_corrlated_feature],
    y=df['target']
)

plt.title(f"{most_corrlated_feature} VS Target")
plt.xlabel(most_corrlated_feature)
plt.ylabel("Target")
plt.show()
```

**obervation:**

    The scatter plot shows the relationship between the most correlated feature and the target variable. The direction of the relationship can be understood from the pattern of the points. A stronger correlation generally results in points following a clearer trend.

**Q4. Data Preprocessing**

```python
X = df[['bmi']]
Y = df[['target']]
display(X)
display(Y)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer

num_pipe = make_pipeline(SimpleImputer(strategy='median')).set_output(transform='pandas')
num_pipe
```

```python
# Splitting the data into 80% training and 20% testing sets
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(X,Y,train_size=0.8,random_state=21)
```

**Q5. Model Training**

```python
print("X Training and Testing size :" , xtrain.shape,xtest.shape)
```

```python
print("Y Training and Testing size :" , ytrain.shape,ytest.shape)
```

```python
num_pipe.fit(xtrain)
```

```python
xtrain_pre = num_pipe.transform(xtrain)
xtest_pre = num_pipe.transform(xtest)
```

```python
# module building
from  sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(xtrain_pre,ytrain)
```

```python
print(f"Yintercept: {model.intercept_},Slope: {model.coef_}")
```

## Q6. Prediction

```python
ypred_test = model.predict(xtest_pre)
ypred_test[:5]
```

**Q7. Model Evaluation**

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score

mse = mean_squared_error(ytest,ypred_test)
rmse = mse**(1/2)
mae = mean_absolute_error(ytest,ypred_test)
r2 =  r2_score(ytest,ypred_test)

print(f"MSE : {mse}")
print(f"RMSE : {rmse}")
print(f"MAE : {mae}")
print(f"R2 : {r2*100:.2f}%")
```
