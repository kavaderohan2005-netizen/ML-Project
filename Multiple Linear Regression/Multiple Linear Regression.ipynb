## Step3: Separate the X and Y features

    X: RND , MKT
    Y: PROFIT

```python
X = df[['RND','MKT']]
Y = df[['PROFIT']]
display(X)
display(Y)
```

### Step4: Feature Engineering:-> Data Cleaning

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer

num_pipe = make_pipeline(SimpleImputer(strategy='median')).set_output(transform='pandas')
num_pipe
```

### Step5: Separate our data into training and testing

dataset size is small: 80% and above training : 20% and below testing

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(X,Y,train_size=0.8,random_state=21)
```

```python
print("X Training and Testing size :" , xtrain.shape,xtest.shape)
```

```python
print("Y Training and Testing size :" , ytrain.shape,ytest.shape)
```

### Apply the pipeline on training and testing data

```python
num_pipe.fit(xtrain)
```

```python
xtrain_pre = num_pipe.transform(xtrain)
xtest_pre = num_pipe.transform(xtest)
```

### Step5: Model Building

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(xtrain_pre,xtrain)
```

### Step 6: Model Evaluation : Performance Metrics

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.linear_model import LinearRegression  # example model
import numpy as np

# Example dataset (replace with your own X, y)
# X = ...
# y = ...

# Split into train and test
X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

# Train model
model = LinearRegression()
model.fit(X_train, y_train)

# Predict on test set only
ypred_test = model.predict(X_test)

# Ensure arrays are 1-D
y_test = np.ravel(y_test)
ypred_test = np.ravel(ypred_test)

# Metrics
mse = mean_squared_error(y_test, ypred_test)
mae = mean_absolute_error(y_test, ypred_test)
rmse = mse**0.5
r2 = r2_score(y_test, ypred_test)

print("Testing scores")
print(f"MSE: {mse}")
print(f"MAE: {mae}")
print(f"RMSE: {rmse}")
print(f"R2 score - Coefficient of Determination: {r2*100:.2f}")
```

### R2 Score for the model is above 80%. We can consider this model for final deployments.
